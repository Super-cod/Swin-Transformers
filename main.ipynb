Patch embedding 

```python
class PatchEmbeddings(nn.Module):
    def __init__(self,img_size=224,patch_size=4,in_channel=3,embed_dem=96):
        super().__init__()
        self.proj=nn.Conv2d(
            in_channels=in_channel,
            out_channels=embed_dem,
            kernel_size=patch_size,
            stride=patch_size
        )


    def forward(self,x):
        x=self.proj(x)   #B,C,H,W-> B, embed_size, h/k , w/k
        x=x.flatten(2).transpose(1,2)  # B, h/k*w/k , embed_size



```
